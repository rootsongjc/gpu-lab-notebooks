## Jupyter Notebook 是什么

**Jupyter Notebook** 是一种把代码、运行结果和说明文字放在同一个文件（`.ipynb`，本质是 JSON）里的交互式环境。它的核心模型只有三条：



| 概念 | 含义 | 对应你熟悉的东西 |
| :--- | :--- | :--- |
| **Cell（单元）** | 文件由一个个 cell 组成，分 **代码 cell** 和 **文本 cell**（Markdown） | 一个代码 cell ≈ 一次 `python -c` 或一段粘进 REPL 的代码 |
| **Kernel（内核）** | 后台常驻的 Python 进程，所有代码 cell 都在**同一个进程**里执行，变量互相可见 | 一个一直开着的 `python` REPL |
| **执行顺序** | 你点哪个 cell 它就跑哪个，顺序由你决定；cell 左边 `[3]` 这种数字是第几次执行 | 在 REPL 里按任意顺序粘代码 |



第三条是最容易踩坑的地方：**cell 在文件里的位置和它的执行顺序无关**。如果你跳过了定义 `a` 的 cell 直接跑用到 `a` 的 cell，会报 `NameError`。本教程的 notebook 都设计成从上到下依次执行即可。

**Google Colab**（Colaboratory）是 Google 托管的 Jupyter：Notebook 文件存在 Google Drive 或 GitHub，Kernel 跑在 Google 的虚拟机上，免费版可以分到一块 Tesla T4。你本地只需要一个浏览器。

### 代码 cell 里的三种"方言"



| 写法 | 含义 | 例子 |
| :--- | :--- | :--- |
| 普通 Python | 交给 kernel 执行 | `import torch` |
| 行首 `!` | 这一行交给 **shell** 执行，相当于在终端敲命令 | `!nvidia-smi`、`!pip install -q nvidia-ml-py` |
| 行首 `%` 或 `%%` | Jupyter **魔法命令**，`%%` 作用于整个 cell | `%%writefile hello.cu` 把 cell 内容写成文件 |



`!` 和 `%%writefile` 是本教程用得最多的两个。注意 `!cd /tmp` 不会改变后续 cell 的工作目录（每个 `!` 是独立的子 shell），要用 `%cd /tmp`。

## 故障模式



| 症状 | 原因 | 处理 |
| :--- | :--- | :--- |
| 右下角显示 `CPU` 或 `Connect` | 还没切 GPU / 还没连接 | Runtime → Change runtime type → T4 GPU → Save |
| 「Cannot connect to GPU backend」 | 高峰期免费 GPU 用尽 | 稍后重试，或先用 CPU runtime 读文字部分 |
| 顶部黄色横幅「connected to a GPU runtime, but not utilizing the GPU」 | 连上 GPU 后一段时间没跑 GPU 代码，Colab 提醒别占着 | 继续做实验即可；点 Close 关掉 |
| `NameError: name 'xxx' is not defined` | 跳过了前面定义它的 cell，或 runtime 重启过 | 从上往下重新运行，或 Runtime → Run all |
| 装好的包在新 cell 里 `ModuleNotFoundError` | runtime 被回收后重连，是一台新机器 | 重新运行安装 cell |
| 修改后关闭页面，再打开全没了 | GitHub 打开的 notebook 是只读的 | 先 Copy to Drive |
| 「Warning: This notebook was not authored by Google」 | 所有非 Google 来源 notebook 都会提示 | 审阅源码后 Run anyway |
